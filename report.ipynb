# Optimality Gap and Computation Time

```python
import pandas as pd
import json
import os
import matplotlib.image as mpimg
import matplotlib.pyplot as plt
```

```python
naive_cols = ["naive_obj",  "naive_road_u", "naive_int_u", "naive_time", "naive_x1_count", "naive_x2_count", "naive_y_count", "naive_policy_sim"]
heu_cols = ["heu_obj",  "heu_road_u", "heu_int_u", "heu_time", "heu_x1_count", "heu_x2_count", "heu_y_count", "heu_policy_sim"]
solver_cols = ["solver_obj",  "solver_road_u", "solver_int_u", "solver_time", "solver_x1_count", "solver_x2_count", "solver_y_count", "solver_policy_sim"]
allRoadsDF = pd.DataFrame(columns = ["scenarioID"] + naive_cols + heu_cols + solver_cols)

for id in range(1, 8):
    with open(f"sol_heuristic/output_h1/scenario_{id}/meta_data.json", "r") as file:
        naive = json.load(file)
    with open(f"sol_heuristic/output_h2/scenario_{id}/meta_data.json", "r") as file:
        heuristic = json.load(file)
    with open(f"solver/output/scenario_{id}/meta_data.json", "r") as file:
        solver = json.load(file)

    allRoadsDF.loc[len(allRoadsDF), "scenarioID"] = id
    allRoadsDF.loc[allRoadsDF["scenarioID"] == id, naive_cols] = [
        naive["obj_val"]["total_utility"],
        naive["obj_val"]["road_utility"],
        naive["obj_val"]["int_utility"],
        naive["cal_time_sec"],
        naive["result_description"]["num_x1"],
        naive["result_description"]["num_x2"],
        naive["result_description"]["num_y"],
        naive["policy_similarity"]
    ]
    allRoadsDF.loc[allRoadsDF["scenarioID"] == id, heu_cols] = [
        heuristic["obj_val"]["total_utility"],
        heuristic["obj_val"]["road_utility"],
        heuristic["obj_val"]["int_utility"],
        heuristic["cal_time_sec"],
        heuristic["result_description"]["num_x1"],
        heuristic["result_description"]["num_x2"],
        heuristic["result_description"]["num_y"],
        heuristic["policy_similarity"]
    ]
    allRoadsDF.loc[allRoadsDF["scenarioID"] == id, solver_cols] = [
        solver["obj_val"]["total_utility"],
        solver["obj_val"]["road_utility"],
        solver["obj_val"]["int_utility"],
        solver["cal_time_sec"],
        solver["result_description"]["num_x1"],
        solver["result_description"]["num_x2"],
        solver["result_description"]["num_y"],
        solver["policy_similarity"]
    ]
```

```python
naive_cols = ["naive_obj",  "naive_road_u", "naive_int_u", "naive_time", "naive_x1_count", "naive_x2_count", "naive_y_count", "naive_policy_sim"]
heu_cols = ["heu_obj",  "heu_road_u", "heu_int_u", "heu_time", "heu_x1_count", "heu_x2_count", "heu_y_count", "heu_policy_sim"]
solver_cols = ["solver_obj",  "solver_road_u", "solver_int_u", "solver_time", "solver_x1_count", "solver_x2_count", "solver_y_count", "solver_policy_sim"]
newLanesDF = pd.DataFrame(columns = ["scenarioID"] + naive_cols + heu_cols + solver_cols)

for id in range(1, 8):
    with open(f"sol_heuristic/output_h1/scenario_{id}_np/meta_data.json", "r") as file:
        naive = json.load(file)
    with open(f"sol_heuristic/output_h2/scenario_{id}_np/meta_data.json", "r") as file:
        heuristic = json.load(file)
    with open(f"solver/output/scenario_{id}_np/meta_data.json", "r") as file:
        solver = json.load(file)

    newLanesDF.loc[len(newLanesDF), "scenarioID"] = id
    newLanesDF.loc[newLanesDF["scenarioID"] == id, naive_cols] = [
        naive["obj_val"]["total_utility"],
        naive["obj_val"]["road_utility"],
        naive["obj_val"]["int_utility"],
        naive["cal_time_sec"],
        naive["result_description"]["num_x1"],
        naive["result_description"]["num_x2"],
        naive["result_description"]["num_y"],
        naive["policy_similarity"]
    ]
    newLanesDF.loc[newLanesDF["scenarioID"] == id, heu_cols] = [
        heuristic["obj_val"]["total_utility"],
        heuristic["obj_val"]["road_utility"],
        heuristic["obj_val"]["int_utility"],
        heuristic["cal_time_sec"],
        heuristic["result_description"]["num_x1"],
        heuristic["result_description"]["num_x2"],
        heuristic["result_description"]["num_y"],
        heuristic["policy_similarity"]
    ]
    newLanesDF.loc[newLanesDF["scenarioID"] == id, solver_cols] = [
        solver["obj_val"]["total_utility"],
        solver["obj_val"]["road_utility"],
        solver["obj_val"]["int_utility"],
        solver["cal_time_sec"],
        solver["result_description"]["num_x1"],
        solver["result_description"]["num_x2"],
        solver["result_description"]["num_y"],
        solver["policy_similarity"]
    ]
```

```python
for col in allRoadsDF.columns:
    allRoadsDF[col] = allRoadsDF[col].astype(float)
allRoadsDF = allRoadsDF.round(3)
allRoadsDF["optgap_naive"] = (allRoadsDF["solver_obj"] - allRoadsDF["naive_obj"]) / allRoadsDF["solver_obj"]
allRoadsDF["optgap_heu"] = (allRoadsDF["solver_obj"] - allRoadsDF["heu_obj"]) / allRoadsDF["solver_obj"]
allRoadsDF["scenarioID"] = allRoadsDF["scenarioID"].astype(int).astype(str)
allRoadsDF.set_index("scenarioID", inplace = True)
```
